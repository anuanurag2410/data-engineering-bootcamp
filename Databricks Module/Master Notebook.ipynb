# Master Notebook - Run Class 2 to 11

Runs every class notebook from Class 2 through Class 11 in order, passing the target Unity Catalog `catalog` as a parameter to each child notebook.

```python


# Base folder = folder that contains this Master Notebook
current_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
base_dir = current_path.rsplit("/", 1)[0]

# Class notebooks to run, in order (there is no Class 3 folder)
class_notebooks = [
    "Class 2/Class 2 Databricks",
    "Class 4/Class_04_Files_Data_Storage",
    "Class 5/Class_05_Databases_Tables_Metastore",
    "Class 6/Class_06_External_Managed_Tables",
    "Class 7/07_Delta_Lake_Fundamentals",
    "Class 8/08_Delta_Time_Travel_Restore_Vacuum",
    "Class 9/09_Delta_Merge_Incremental_Loading",
    "Class 10/10_Unity_Catalog_Fundamentals",
    "Class 11/11_Unity_Catalog_Volumes",
]

print(f"Catalog : {catalog}")
print(f"Base dir: {base_dir}")
for nb in class_notebooks:
    print("  -", f"{base_dir}/{nb}")
```

```python
import time

TIMEOUT_SECONDS = 3600
results = []

for nb in class_notebooks:
    path = f"{base_dir}/{nb}"
    start = time.time()
    try:
        output = dbutils.notebook.run(path, TIMEOUT_SECONDS, {"catalog": catalog})
        status, error = "SUCCESS", None
    except Exception as e:
        output, status, error = None, "FAILED", str(e)[:500]
    duration = round(time.time() - start, 1)
    print(f"{status:<8} {duration:>7}s  {nb}")
    results.append((nb, status, duration, output, error))

display(spark.createDataFrame(
    results,
    "notebook string, status string, duration_sec double, output string, error string",
))
```
